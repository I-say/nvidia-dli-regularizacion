## 1. Listas — muchos valores en una sola variable

Hasta ahora, cada variable guardaba **un solo valor**.  
Una **lista** guarda **muchos valores** en orden, dentro de corchetes `[]`.

````python
# Una lista de nombres de carreras
carreras = ["IA y Ciencia de Datos", "Ciberseguridad", "Creatividad Digital", "Tecnologías Biomédicas", "Inteligencia Financiera"]

# Una lista de calificaciones
calificaciones = [8.5, 9.0, 7.5, 10.0, 8.0]

# Una lista de número de alumnos por carrera
alumnos_por_carrera = [60, 45, 38, 52, 41]

print(carreras)
print(calificaciones)
print(f"\nTotal de carreras en CUGDL: {len(carreras)}")  # len() cuenta los elementos
````

````python
# Acceder a elementos individuales — los índices empiezan en 0
# [0] es el primero, [1] el segundo, [-1] el último

print(carreras[0])   # primer elemento
print(carreras[1])   # segundo elemento
print(carreras[-1])  # último elemento

# En IA, esto sería como acceder a la primera imagen de un dataset
imagenes = ["gato001.jpg", "perro001.jpg", "pajaro001.jpg"]
print(f"\nPrimera imagen del dataset: {imagenes[0]}")
print(f"Última imagen del dataset: {imagenes[-1]}")
````

````python
# Agregar elementos a una lista con .append()
etiquetas = ["gato", "perro", "pájaro"]
print(f"Antes: {etiquetas}")

etiquetas.append("pez")
print(f"Después de append: {etiquetas}")

# Esto es lo que hace Python cuando carga un dataset:
# va agregando imagen por imagen a una lista
dataset = []
dataset.append("imagen_001")
dataset.append("imagen_002")
dataset.append("imagen_003")
print(f"\nDataset construido: {dataset}")
````

## 2. Ciclos `for` — repetir sin aburrirse

Un ciclo `for` le dice a Python:  
**"Para cada elemento en esta lista, haz esto"**

Es la forma de evitar copiar y pegar código mil veces.

````python
# Recorrer una lista con for
frutas = ["manzana", "naranja", "mango", "fresa"]

for fruta in frutas:
    # Todo lo que está indentado (con espacios) se repite
    print(f"Procesando: {fruta}")

print("¡Listo! Se procesaron todas las frutas")
````

````python
# range() genera una secuencia de números
# range(5) genera: 0, 1, 2, 3, 4

# Simulamos el entrenamiento de un modelo por 5 épocas
total_epocas = 5
error = 1.0  # empieza con error máximo

for epoca in range(total_epocas):
    error = error * 0.7   # en cada época el error baja un 30%
    error = round(error, 4)  # redondeamos a 4 decimales
    print(f"Época {epoca + 1}/{total_epocas} — Error: {error}")

print(f"\nEntrenamiento completo. Error final: {error}")
````

````python
# enumerate() — cuando necesitas el índice Y el valor al mismo tiempo
# Muy usado en el código del NVIDIA DLI

carreras = ["IA y Ciencia de Datos", "Ciberseguridad", "Creatividad Digital"]

for indice, carrera in enumerate(carreras):
    print(f"Carrera {indice}: {carrera}")

# Ejemplo más real: recorrer imágenes y sus etiquetas
imagenes = ["img001.jpg", "img002.jpg", "img003.jpg"]
etiquetas = ["fruta_fresca", "fruta_podrida", "fruta_fresca"]

print("\nDataset de frutas:")
for i, imagen in enumerate(imagenes):
    print(f"  [{i}] {imagen} → {etiquetas[i]}")
````

## 3. Condicionales `if` — tomar decisiones

Un `if` le dice a Python:  
**"Si esta condición es verdadera, haz esto. Si no, haz aquello"**

````python
# if básico
calificacion = 8.5

if calificacion >= 9.0:
    print("Excelente — obtuviste certificado con distinción")
elif calificacion >= 7.0:
    print("Aprobado — obtuviste el certificado NVIDIA")
else:
    print("Necesitas repasar — puedes volver a intentarlo")

# Las comparaciones en Python:
# ==  igual a
# !=  diferente de
# >   mayor que
# <   menor que
# >=  mayor o igual
# <=  menor o igual
````

````python
# Ejemplo con contexto de IA:
# Un clasificador de frutas decide si está fresca o podrida
# basándose en un número de confianza (entre 0 y 1)

prediccion = 0.82  # el modelo está 82% seguro de que es fruta fresca

if prediccion >= 0.5:
    resultado = "fresca"
else:
    resultado = "podrida"

print(f"Confianza del modelo: {prediccion * 100:.1f}%")
print(f"Clasificación: la fruta está {resultado}")

# Esto es EXACTAMENTE lo que hace el proyecto final del NVIDIA DLI
````

## 4. Funciones — código reutilizable

Una **función** es como una receta:  
Le das ingredientes (parámetros), sigue instrucciones, y te devuelve un resultado.

```python
def nombre_de_la_funcion(parametro1, parametro2):
    # instrucciones
    return resultado
```

> 💡 En el NVIDIA DLI vas a usar funciones todo el tiempo:  
> `model.fit()`, `model.predict()`, `ImageDataGenerator()` — todas son funciones que alguien ya escribió por ti.

````python
# Función básica — sin parámetros, sin return
def saludar():
    print("¡Hola! Bienvenido al curso de Deep Learning")

# Para usar la función, la llamamos así:
saludar()
saludar()  # podemos usarla las veces que queramos
````

````python
# Función con parámetros — recibe información
def saludar_alumno(nombre, carrera):
    print(f"¡Hola {nombre}! Nos alegra que te unas desde {carrera}")

saludar_alumno("Isay", "IA y Ciencia de Datos")
saludar_alumno("Ana", "Ciberseguridad")
saludar_alumno("Luis", "Creatividad Digital")
````

````python
# Función con return — devuelve un resultado que puedes guardar
def calcular_promedio(calificaciones):
    total = sum(calificaciones)   # sum() suma todos los elementos
    cantidad = len(calificaciones) # len() cuenta los elementos
    promedio = total / cantidad
    return promedio

mis_calificaciones = [8.5, 9.0, 7.5, 10.0, 8.0]
resultado = calcular_promedio(mis_calificaciones)
print(f"Promedio: {resultado}")

# Lo más importante: una función puede llamar a otra función
print(f"Promedio redondeado: {round(resultado, 2)}")
````

````python
# Función con contexto de IA:
# Simula lo que hace un clasificador muy simple

def clasificar_fruta(confianza_frescura):
    """
    Clasifica una fruta basándose en el nivel de confianza del modelo.
    
    Parámetros:
        confianza_frescura (float): número entre 0.0 y 1.0
        
    Regresa:
        str: "fresca" o "podrida"
    """
    if confianza_frescura >= 0.5:
        return "fresca"
    else:
        return "podrida"

# Probamos la función con diferentes valores
predicciones = [0.92, 0.18, 0.67, 0.34, 0.88]

for i, pred in enumerate(predicciones):
    clasificacion = clasificar_fruta(pred)
    print(f"Fruta #{i+1} — confianza: {pred:.0%} → {clasificacion}")
````
