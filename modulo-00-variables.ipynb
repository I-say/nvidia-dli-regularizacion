## 1. ¿Qué es una variable?

Una **variable** es como una caja con nombre donde guardas información.

```
 nombre_de_la_caja  =  lo que guardas adentro
```

El signo `=` en Python **no** significa "igual" como en matemáticas.  
Significa **"guarda esto aquí"**.

Vamos a crear nuestras primeras variables:

````python
# El símbolo # es un comentario — Python lo ignora completamente
# Los comentarios son para que TÚ (o tu equipo) entiendan el código

# Guardamos el nombre de un estudiante
nombre = "Isay"

# Guardamos una edad
edad = 27

# Guardamos una calificación con decimales
calificacion = 9.5

# Guardamos si alguien ya se inscribió al NVIDIA DLI
inscrito = True

# print() muestra el valor de una variable en pantalla
print(nombre)
print(edad)
print(calificacion)
print(inscrito)
````

---
## 2. Tipos de datos — los 4 básicos que necesitas

| Tipo | ¿Qué es? | Ejemplo | En IA se usa para... |
|------|----------|---------|---------------------|
| `int` | Número entero | `5`, `100`, `-3` | Número de épocas, tamaño de batch |
| `float` | Número con decimales | `9.5`, `0.001` | Tasa de aprendizaje, precisión del modelo |
| `str` | Texto (string) | `"hola"`, `"imagen.jpg"` | Nombres de archivos, etiquetas |
| `bool` | Verdadero o falso | `True`, `False` | ¿Está entrenando? ¿Hay GPU disponible? |

La función `type()` te dice qué tipo de dato es algo:

````python
# Verificamos el tipo de cada variable
print(type(nombre))       # str — texto
print(type(edad))         # int — entero
print(type(calificacion)) # float — decimal
print(type(inscrito))     # bool — verdadero/falso
````

---
## 3. Print con estilo — mostrar información útil

En el workshop de NVIDIA vas a ver mucho código que imprime el progreso del entrenamiento.  
Saber usar `print()` bien te ayuda a entender qué está pasando.

Hay varias formas de imprimir:

````python
# Forma 1 — concatenar con coma (la más simple)
print("Nombre:", nombre)
print("Edad:", edad)

# Forma 2 — f-strings (la más usada en código profesional)
# Ponemos f antes de las comillas y usamos {} para insertar variables
print(f"Hola, me llamo {nombre} y tengo {edad} años")
print(f"Mi calificación es: {calificacion}")

# Ejemplo real de cómo se ve en el NVIDIA DLI:
epochs = 5
accuracy = 0.87
print(f"Época {epochs} completada — precisión del modelo: {accuracy}")
````

---
## 4. Operaciones matemáticas

Python es una calculadora muy poderosa. Estas son las operaciones básicas:

````python
# Las 4 básicas
print(10 + 3)   # suma → 13
print(10 - 3)   # resta → 7
print(10 * 3)   # multiplicación → 30
print(10 / 3)   # división → 3.333...

# Dos que vas a ver mucho en IA:
print(10 // 3)  # división entera (sin decimales) → 3
print(10 % 3)   # módulo (el residuo de la división) → 1
print(2 ** 8)   # potencia (2 elevado a la 8) → 256
````

````python
# Ejemplo práctico con contexto de IA:
# Si tienes 1000 imágenes y las procesas en grupos de 32 (batch_size)
# ¿cuántos grupos necesitas?

total_imagenes = 1000
batch_size = 32

grupos = total_imagenes // batch_size
sobrantes = total_imagenes % batch_size

print(f"Total de imágenes: {total_imagenes}")
print(f"Tamaño de cada grupo (batch): {batch_size}")
print(f"Número de grupos completos: {grupos}")
print(f"Imágenes en el último grupo incompleto: {sobrantes}")
````

---
## 5. Actualizar variables

Las variables pueden cambiar su valor. Esto es clave para entender cómo funciona el entrenamiento de un modelo — en cada época, los valores cambian.

````python
# Las variables se pueden actualizar
puntos = 0
print(f"Puntos iniciales: {puntos}")

puntos = puntos + 10   # le sumamos 10
print(f"Después de primer ejercicio: {puntos}")

puntos += 15           # forma corta de decir puntos = puntos + 15
print(f"Después de segundo ejercicio: {puntos}")

# Esto es exactamente lo que hace un modelo cuando "aprende":
# en cada época, el error se actualiza para ser un poco menor
error = 1.0
print(f"\nError inicial del modelo: {error}")

# Simulamos 3 épocas de entrenamiento (muy simplificado)
error -= 0.2   # época 1 → el modelo mejora
print(f"Después de época 1: {error}")

error -= 0.15  # época 2 → sigue mejorando
print(f"Después de época 2: {error}")

error -= 0.1   # época 3 → mejora un poco menos
print(f"Después de época 3: {error}")
````
