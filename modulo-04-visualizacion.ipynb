````python
import numpy as np
import matplotlib.pyplot as plt

# Configuración para que las gráficas se vean bien
plt.rcParams['figure.figsize'] = (10, 4)
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

print("Librerías listas")
````

## 1. Gráfica básica de línea

La gráfica más común en Deep Learning: una curva que muestra cómo evoluciona una métrica.

````python
# Simulamos el error de un modelo durante 20 épocas de entrenamiento
epocas = list(range(1, 21))  # [1, 2, 3, ..., 20]

# El error baja con ruido (como en la realidad)
np.random.seed(42)  # para resultados reproducibles
error = [1.0 * (0.85 ** e) + np.random.uniform(-0.02, 0.02) for e in range(20)]

# Crear la gráfica
plt.figure(figsize=(10, 4))
plt.plot(epocas, error, color='#d62728', linewidth=2, marker='o', markersize=4)
plt.xlabel('Época')
plt.ylabel('Error (loss)')
plt.title('Curva de entrenamiento — el error baja con cada época')
plt.tight_layout()
plt.show()

print(f"Error inicial: {error[0]:.3f}")
print(f"Error final:   {error[-1]:.3f}")
print(f"Mejora: {(1 - error[-1]/error[0]) * 100:.1f}%")
````

---
## 2. Las curvas del NVIDIA DLI — la gráfica más importante

Durante el workshop vas a ver esta gráfica constantemente.  
Tiene **dos líneas**: entrenamiento y validación.

- **Training accuracy** → qué tan bien clasifica las imágenes con las que entrena
- **Validation accuracy** → qué tan bien clasifica imágenes que NUNCA había visto

La validación es la que importa — mide si el modelo realmente aprendió o solo memorizó.

````python
# Simulamos las curvas de entrenamiento típicas del NVIDIA DLI
epocas = list(range(1, 26))
np.random.seed(0)

# Train accuracy sube rápido
train_acc = [min(0.99, 0.5 + 0.47 * (1 - np.exp(-e/3)) + np.random.uniform(-0.01, 0.01)) for e in range(25)]

# Val accuracy sube más lento y se estabiliza
val_acc = [min(0.95, 0.5 + 0.40 * (1 - np.exp(-e/5)) + np.random.uniform(-0.02, 0.02)) for e in range(25)]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Gráfica de accuracy
axes[0].plot(epocas, train_acc, label='Entrenamiento', color='#1f77b4', linewidth=2)
axes[0].plot(epocas, val_acc, label='Validación', color='#ff7f0e', linewidth=2, linestyle='--')
axes[0].set_xlabel('Época')
axes[0].set_ylabel('Precisión (accuracy)')
axes[0].set_title('Curvas de precisión')
axes[0].legend()
axes[0].set_ylim([0.4, 1.0])

# Gráfica de loss (error)
train_loss = [max(0.05, 0.8 * np.exp(-e/4) + np.random.uniform(-0.02, 0.02)) for e in range(25)]
val_loss   = [max(0.1,  0.8 * np.exp(-e/6) + np.random.uniform(-0.03, 0.03)) for e in range(25)]

axes[1].plot(epocas, train_loss, label='Entrenamiento', color='#1f77b4', linewidth=2)
axes[1].plot(epocas, val_loss, label='Validación', color='#ff7f0e', linewidth=2, linestyle='--')
axes[1].set_xlabel('Época')
axes[1].set_ylabel('Error (loss)')
axes[1].set_title('Curvas de error')
axes[1].legend()

plt.suptitle('Lo que verás en el NVIDIA DLI Workshop', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
````

````python
# Cómo leer las curvas:
print("=" * 55)
print("CÓMO LEER LAS CURVAS DEL NVIDIA DLI")
print("=" * 55)
print()
print(" BUEN modelo:")
print("   - Train y Validation suben juntas")
print("   - El gap entre ellas es pequeño")
print("   - Ambas se estabilizan al final")
print()
print("  OVERFITTING (el peor enemigo):")
print("   - Train sigue subiendo")
print("   - Validation se estanca o baja")
print("   - El modelo memorizó, no aprendió")
print()
print("Para evitar overfitting, el workshop usa:")
print("   → Data augmentation (módulo 3 del workshop)")
print("   → Dropout layers")
print("   → Transfer learning")
````

## 3. Visualizar imágenes con matplotlib

Así es como el workshop muestra imágenes del dataset.

````python
# Crear imágenes sintéticas para visualizar
# (en el workshop real usarás imágenes de frutas)

# Imagen 'fresca' — tonos verdes/brillantes
np.random.seed(1)
img_fresca = np.random.randint(100, 255, size=(32, 32, 3), dtype=np.uint8)
img_fresca[:, :, 0] = np.random.randint(50, 120, size=(32, 32))   # R bajo
img_fresca[:, :, 1] = np.random.randint(150, 255, size=(32, 32))  # G alto → verde
img_fresca[:, :, 2] = np.random.randint(50, 100, size=(32, 32))   # B bajo

# Imagen 'podrida' — tonos café/oscuros
img_podrida = np.random.randint(30, 150, size=(32, 32, 3), dtype=np.uint8)
img_podrida[:, :, 0] = np.random.randint(100, 180, size=(32, 32)) # R medio
img_podrida[:, :, 1] = np.random.randint(60, 120, size=(32, 32))  # G bajo → café
img_podrida[:, :, 2] = np.random.randint(20, 60, size=(32, 32))   # B muy bajo

# Visualizar las dos imágenes
fig, axes = plt.subplots(1, 2, figsize=(8, 4))

axes[0].imshow(img_fresca)
axes[0].set_title('Fruta Fresca\n(tonos verdes)', fontsize=12)
axes[0].axis('off')

axes[1].imshow(img_podrida)
axes[1].set_title('Fruta Podrida\n(tonos café/oscuros)', fontsize=12)
axes[1].axis('off')

plt.suptitle('Así ve el modelo las imágenes (como matrices de números RGB)', y=1.02)
plt.tight_layout()
plt.show()

print(f"\nForma de cada imagen: {img_fresca.shape}")
print(f"Tipo de datos: {img_fresca.dtype}")
print(f"\nPíxel [0,0] de la fruta fresca (R, G, B):  {img_fresca[0,0]}")
print(f"Píxel [0,0] de la fruta podrida (R, G, B): {img_podrida[0,0]}")
````

## 4. Visualizar un batch de imágenes

En el workshop, el modelo procesa imágenes en grupos (batches).  
Así se visualiza un batch:

````python
# Simular un batch de 8 imágenes (4 frescas, 4 podridas)
etiquetas_batch = ["fresca", "podrida", "fresca", "fresca",
                   "podrida", "fresca", "podrida", "podrida"]

fig, axes = plt.subplots(2, 4, figsize=(12, 6))
axes = axes.flatten()

for i, etiqueta in enumerate(etiquetas_batch):
    np.random.seed(i * 7)
    if etiqueta == "fresca":
        img = np.zeros((32, 32, 3), dtype=np.uint8)
        img[:, :, 0] = np.random.randint(50, 120, (32, 32))
        img[:, :, 1] = np.random.randint(150, 255, (32, 32))
        img[:, :, 2] = np.random.randint(50, 100, (32, 32))
        color_titulo = 'green'
    else:
        img = np.zeros((32, 32, 3), dtype=np.uint8)
        img[:, :, 0] = np.random.randint(100, 180, (32, 32))
        img[:, :, 1] = np.random.randint(60, 120, (32, 32))
        img[:, :, 2] = np.random.randint(20, 60, (32, 32))
        color_titulo = 'brown'
    
    axes[i].imshow(img)
    axes[i].set_title(etiqueta, color=color_titulo, fontsize=11, fontweight='bold')
    axes[i].axis('off')

plt.suptitle(f'Batch de {len(etiquetas_batch)} imágenes — así las procesa el modelo', fontsize=12, y=1.02)
plt.tight_layout()
plt.show()
````
